```python
#Task Manager program

tasks= {}

while True:
    print("Personal Task Manager \n")
    print("1. Add Task \n")
    print("2. View Tasks \n")
    print("3. Remove Task \n")
    print("4. Save Tasks to File \n")
    print("5. Load Tasks from File \n")
    print("6. Exit \n")

    choice = input("choose an option \n")
    if choice == 1:
      task = input("enter the task: \n")
      priority = input("Enter the priority High/Medium/Low \n")
      tasks[task] = priority
    elif choice == 2:
      print("Tasks: \n")
      for task,priority in sorted(tasks.items()):
          print(f" - {task} [{priority}]")
    elif choice == 3:
      task = input("Enter the task to remove \n")
      tasks.remove(task,none)
    elif choice == 4:
        with open(tasks.txt,"w") as file:
          for task,priority in tasks.items():
            file.write(f"{task} - [{priority}] \n")
    elif choice == 5:
      with open("tasks.txt", "r")  as file:
          for line in file:
            task , priority = line.strip().split(",")
            tasks[task] =  priority
    elif choice == 6:
      break
```
